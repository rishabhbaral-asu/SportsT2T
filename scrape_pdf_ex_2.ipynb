```python
import re

def extract_text_between_overs(text):
    pattern = r"\b([0-4]?[0-9]\.[1-6])\b"  # Matches over markers (e.g., 0.1 - 49.6)
    matches = list(re.finditer(pattern, text))  # Find all over markers with positions
    
    extracted_sections = []
    
    for i in range(len(matches)-1):
        start = matches[i].start()  # End position of current over
        end = matches[i + 1].start()  # Start position of next over
        extracted_sections.append(text[start:end].strip())  # Extract text in between

    return extracted_sections
```

```python
def get_overs(comm:list):
    try:
        return float(comm[0])
    except:
        return 0.0

def get_runs(comm:list):
    team_runs = 0
    run = comm[1].strip()
    if run == '•':
        run = '0'
    if run[0].isdigit()==True:
        team_runs = float(run[0])
    return run, team_runs

"""def get_commentary(comm:list, run, batter):
    dismissal = [""]
    if 'W' in run:
        for c in comm:
            d = ""
            if batter in c:
                ind = c.find(batter)
                search = c[ind+len(batter):].split(" ")
                for s in search:
                    if s.isdigit()==True:
                        break
                    d = d+" "+s
            if d!="":
                dismissal.append(d)
    commentary = ' '.join(comm[2:8])
    return commentary, dismissal[-1].strip()"""

"""def get_commentary(comm:list, run, batter):
    dismissals = [" st "," c "," run out "," lbw b "," hit wicket "," obstructing the field "," retired out "," b "]
    dismissal = [""]
    commentary = ' '.join(comm[2:10])
    batter = batter.split(" ")[-1]
    #print(commentary)
    if 'W' in run:
        match = next(((word, commentary.index(word)) for word in dismissals if word in commentary), None)
        d = ""
        if match:
            #print(batsman_find)
            search = commentary[match[1]:].split(" ")
            d = ""
            for s in search:
                if s.isdigit()==True:
                    break
                d = d+" "+s
        if d!="":
            dismissal.append(d)
    return commentary, dismissal[-1].strip()"""

def get_commentary(comm:list, run, batter):
    dismissals = [" st "," c "," run out "," lbw b "," hit wicket "," obstructing the field "," retired out "," b "]
    dismissal = [""]
    commentary = ' '.join(comm[2:])
    commentary_final = ""
    for c in comm[2:]:
        if "\"" in c or re.findall(r'\b\d{2}(?:am|pm)\b', c, re.IGNORECASE):
            break
        if 'CRR:' in c or 'end of over' in c:
            break
        commentary_final = commentary_final+" "+c
        if ' SR: ' in c or 'end of over' in c:
            break
    #batter = batter.split(" ")[-1]
    #print(commentary)
    if 'W' in run:
        r_ind = commentary_final.rfind(batter)
        match = next(((word, commentary_final[r_ind:].index(word)) for word in dismissals if word in commentary_final[r_ind:]), None)
        print(match," ",r_ind)
        commentary_2 = commentary_final[r_ind:]
        d = ""
        if match:
            #print(batsman_find)
            search = commentary_2[match[1]:].split(" ")
            d = ""
            for s in search:
                if s.isdigit()==True:
                    break
                d = d+" "+s
        if d!="":
            dismissal.append(d)
    return commentary_final, dismissal[-1].strip()

'''def get_commentary(comm:list, run, batter):
    dismissals = [" st "," c "," run out "," lbw "," hit wicket "," obstructing the field "," retired out "," b "]
    dismissal = [""]
    commentary = ""
    for ele in comm[2:10]:
        if 'CRR:' in ele or 'end of over' in ele:
            break
        commentary = commentary+" "+ele
    batter = batter.split(" ")[-1]
    #print(commentary)
    if 'W' in run:
        match = next(((word, commentary.index(word)) for word in dismissals if word in commentary), None)
        d = ""
        if match:
            #print(batsman_find)
            search = commentary[match[1]:].split(" ")
            d = ""
            for s in search:
                if s.isdigit()==True:
                    break
                d = d+" "+s
        if d!="":
            dismissal.append(d)
    return commentary, dismissal[-1].strip()'''

def get_batter_bowler(comm:list):
    search = comm[2]
    s2 = search.find(' to ')
    s3 = search.find(',')
    batter = search[s2+3:s3].strip()
    bowler = search[:s2].strip()
    return bowler, batter

def get_runs_batter(run):
    if run.isdigit()==True:
        return float(run)
    elif run.isdigit()==False and 'nb' in run and run[0].isdigit():
        return float(run[0])-1
    else:
        return 0

def get_4_batter(run):
    if run.isdigit()==True and run=='4':
        return 1
    elif run.isdigit()==False and run[0]=='4' and 'lb' not in run:
        return 1
    else:
        return 0

def get_6_batter(run):
    if run.isdigit()==True and run=='6':
        return 1
    elif run.isdigit()==False and run[0]=='6' and 'lb' not in run:
        return 1
    else:
        return 0

def get_bowls_batter(run):
    if run.isdigit()==True or run=='W':
        return 1
    elif run.isdigit()==False and 'w' not in run:
        return 1
    else:
        return 0

def get_bowls_bowler(run):
    if run.isdigit()==True:
        return 1
    elif 'w' not in run and 'nb' not in run:
        return 1
    else:
        return 0
    
def get_runs_bowler(run):
    if run.isdigit()==True:
        return float(run)
    elif run[0].isdigit()==True and ('nb' in run or 'w' in run):
        return float(run[0])
    else:
        return 0

def get_wickets_bowler(run,commentary):
    dismissals = [" st "," c "," lbw "," hit wicket "," b "]
    if 'W' in run:
        match = next(((word, commentary.index(word)) for word in dismissals if word in commentary), None)
        if match:
            return 1
        else:
            return 0
    else:
        return 0
```

```python
overs = []
runs = []
commentaries = []
commentary_raw = []
batsmen = []
bowlers = []
batter_run_current = []
batter_bowl_current = []
batter_6_current = []
batter_4_current = []
bowler_bowl_current = []
bowler_runs_current = []
bowler_wicket_current = []
dismissal_current = []
team_runs_current = []
batsman_runs = 0
batsman_6 = 0
batsman_4 = 0
batsman_bowls = 0
bowler_runs = 0
bowler_bowls = 0
bowler_wickets = 0
for comm in ball_by_ball_raw:
    #print(comm)
    split_by_spaces = comm.split(' ')
    split_by_line = comm.split('\n')
    over = get_overs(split_by_spaces)
    if len(overs)>0 and over<overs[-1]:
        continue
    if len(overs)>0:
        if ('w' in runs[-1] or 'nb' in runs[-1]):
            next_ball = int(overs[-1])*6+int((overs[-1]*10)%10)
        else:
            next_ball = int(overs[-1])*6+int((overs[-1]*10)%10)+1
        curr_ball = int(over)*6+int((over*10)%10)
        if curr_ball!=next_ball:
            print(comm)
            continue
    if str(over)+' '+'\n' not in comm:
        continue
    run, team_runs = get_runs(split_by_line)
    bowler, batter = get_batter_bowler(split_by_line)
    commentary, dismissal = get_commentary(split_by_line,run,batter)

    ## getting runs by batter
    batsman_runs = get_runs_batter(run)

    ## getting 4 by batter
    batsman_4 = get_4_batter(run)
    
    ## getting 6 by batter
    batsman_6 = get_6_batter(run)

    ## getting bowls faced by batter
    batsman_bowls = get_bowls_batter(run)

    ## getting bowls bowled by bowler
    bowler_bowls = get_bowls_bowler(run)

    ## getting runs given by bowler
    bowler_runs = get_runs_bowler(run)

    ## getting wickets taken by bowler
    bowler_wickets = get_wickets_bowler(run,commentary)
    
    commentary_raw.append(comm)
    overs.append(over)
    runs.append(run)
    team_runs_current.append(team_runs)
    commentaries.append(commentary)
    bowlers.append(bowler)
    batsmen.append(batter)
    batter_run_current.append(batsman_runs)
    batter_4_current.append(batsman_4)
    batter_6_current.append(batsman_6)
    batter_bowl_current.append(batsman_bowls)
    bowler_bowl_current.append(bowler_bowls)
    bowler_runs_current.append(bowler_runs)
    bowler_wicket_current.append(bowler_wickets)
    dismissal_current.append(dismissal)

data = {'raw_data':commentary_raw,
        'overs':overs,
        'runs':runs,
        'team_runs':team_runs_current,
        'commentary':commentaries,
        'bowler':bowlers,
        'batsman':batsmen,
        'batsman_runs':batter_run_current,
        'batsman_fours':batter_4_current,
        'batsman_sixes':batter_6_current,
        'batsman_bowls_faced':batter_bowl_current,
        'bowler_bowls_done': bowler_bowl_current,
        'bowler_runs_given': bowler_runs_current,
        'bowler_wickets': bowler_wicket_current,
        'dismissal':dismissal_current}

df = pd.DataFrame(data)
df
```

```python
# Function to count maidens
def count_maidens(df, bowler):
    maiden_overs = 0
    over_groups = df[df["bowler"] == bowler].groupby(df["overs"].astype(int))  # Group by over number
    for _, over_df in over_groups:
        if over_df["runs_given_bool"].sum() == 0:  # If no runs were given in the over
            maiden_overs += 1
    return maiden_overs
```
