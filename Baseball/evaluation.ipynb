# Part 1: Loading all language files (PbP) into separate files for each language in preparation of creation of cohesive dataset

```python
# --- Configuration ---
# Define the directory where your JSON files are located
json_directory = 'output'

# --- Data Storage ---
# List to store all loaded JSON data, regardless of language
all_json_data = []

# Dictionaries to store data specifically by language
spanish_json_data = []
english_json_data = []
dutch_json_data = []
# You can add more lists here for other languages if needed

# List to store full paths of all found JSON files
file_paths = []

# --- 1. Locate and Filter JSON Files ---
print(f"Searching for JSON files in: '{json_directory}'...")

# Check if the directory exists
if not os.path.exists(json_directory):
    print(f"Error: Directory '{json_directory}' not found. Please ensure it exists and contains your JSON files.")
else:
    # List all files in the specified directory
    for filename in os.listdir(json_directory):
        # Filter for JSON files that match the common naming pattern
        if filename.endswith(".json") and filename.startswith("mlb_play_n_score_"):
            full_file_path = os.path.join(json_directory, filename)
            file_paths.append(full_file_path)

    # Sort file paths to ensure consistent processing order (e.g., by date in filename)
    file_paths.sort()

    print(f"Found {len(file_paths)} relevant JSON files:")
    for path in file_paths:
        print(f"- {path}")

# --- 2. Load and Categorize JSON Data ---
if file_paths:
    print("\nLoading and categorizing JSON data...")
    for file_path in file_paths:
        try:
            # Open file with UTF-8 encoding for robustness
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                all_json_data.append(data)

                # Categorize data by language based on filename
                if '_spanish_' in file_path:
                    spanish_json_data.append(data)
                elif '_english_' in file_path:
                    english_json_data.append(data)
                elif '_dutch_' in file_path:
                    dutch_json_data.append(data)
                # Add more `elif` conditions here for other languages if they exist in your files

        except json.JSONDecodeError as e:
            print(f"Warning: Could not decode JSON from '{file_path}'. Error: {e}. Skipping this file.")
        except FileNotFoundError:
            print(f"Error: File not found at '{file_path}'. This shouldn't happen after `os.listdir`, but good for robust handling. Skipping.")
        except Exception as e:
            print(f"An unexpected error occurred while reading '{file_path}'. Error: {e}. Skipping this file.")
else:
    print("No JSON files found matching the criteria. Nothing to load.")

# --- 3. Summary of Loaded Data ---
print("\n--- Loading Summary ---")
print(f"Total JSON objects loaded: {len(all_json_data)}")
print(f"Spanish JSON objects loaded: {len(spanish_json_data)}")
print(f"English JSON objects loaded: {len(english_json_data)}")
print(f"Dutch JSON objects loaded: {len(dutch_json_data)}")
# --- 4. Merge Data ---
# Save each language's data into a single JSON file
json.dump(spanish_json_data, open('spanish_data.json', 'w', encoding='utf-8'), ensure_ascii=False, indent=4)
json.dump(english_json_data, open('english_data.json', 'w', encoding='utf-8'), ensure_ascii=False, indent=4)
json.dump(dutch_json_data, open('dutch_data.json', 'w', encoding='utf-8'), ensure_ascii=False, indent=4)
```

## 1. Temporal Commentary

```python
# --- a. Take the input and parse it by looking for the key 'top' or 'bottom' ---
def parse_innings(input_string):
    """
    Parses the input string to extract innings data, organizing it by inning number.

    Args:
        input_string (str): The input string containing innings data.

    Returns:
        dict: A dictionary where keys are inning numbers (1-9) and values are lists
              containing the "Top" and/or "Bottom" data for that inning.
    """
    innings_data = {str(i): [] for i in range(1, 10)}  # Initialize 9 lists for innings 1-9
    lines = input_string.split('\n\n')

    for line in lines:
        # Remove leading/trailing dashes and spaces, then split by space
        cleaned_line = line.strip(' -').strip()
        parts = cleaned_line.split(' ')

        if len(parts) >= 2: # Ensure there are at least two parts (e.g., "Top" and "1st")
            inning_num_str_raw = parts[1] # "1st", "2nd", etc.

            # Extract just the number from "1st", "2nd", etc.
            inning_num_str = ''.join(filter(str.isdigit, inning_num_str_raw))

            if inning_num_str and inning_num_str in innings_data:
                innings_data[inning_num_str].append(line)

    return innings_data
# --- b. Apply the parsing function to the English input ---
parsed_innings = [parse_innings(input_str) for input_str in english_input]
# --- c. Convert the parsed innings into a DataFrame ---
innings_df = pd.DataFrame(parsed_innings)
# --- d. Display the innings DataFrame ---
# --- e. Add the 'game_id' column to the innings DataFrame ---
innings_df['game_id'] = merged_df['game_id'].values
# --- f. Reorder the columns to have 'game_id' first ---
innings_df = innings_df[['game_id'] + [col for col in innings_df.columns if col != 'game_id']]
```

```python
# --- h. merge the innings DataFrame with the merged DataFrame ---
merged_df = pd.merge(merged_df, innings_df, on='game_id', how='outer')
```

```python
import re

def anonymize_baseball_commentary(commentary: str) -> str:
    """
    Anonymizes baseball commentary by replacing player names with Player <number>,
    pitcher names with Pitcher <number>, and specific team references with
    TEAM A or TEAM B. It also formats the output with consistent line breaks.

    Args:
        commentary: The raw baseball commentary string.

    Returns:
        The anonymized and formatted commentary string.
    """

    # Start with the original commentary
    anonymized_commentary = commentary

    # 1. Initial Team Anonymization (done on the initial commentary string)
    # Be specific to avoid replacing parts of player names if they contain "reds" or "diamondbacks"
    anonymized_commentary = re.sub(r'Diamondbacks\s*ARI', 'TEAM A', anonymized_commentary)
    anonymized_commentary = re.sub(r'Diamondbacks', 'TEAM A', anonymized_commentary)
    anonymized_commentary = re.sub(r'Reds\s*CIN', 'TEAM B', anonymized_commentary)
    anonymized_commentary = re.sub(r'Reds', 'TEAM B', anonymized_commentary)
    anonymized_commentary = re.sub(r'\(ARI\)', '(TEAM A)', anonymized_commentary)
    anonymized_commentary = re.sub(r'\(CIN\)', '(TEAM B)', anonymized_commentary)
    anonymized_commentary = re.sub(r'ARI\.', 'TEAM A.', anonymized_commentary) # For "PITCHING FOR ARI."
    anonymized_commentary = re.sub(r'CIN\.', 'TEAM B.', anonymized_commentary) # For "PITCHING FOR CIN."

    # 2. Collect all potential player/pitcher names for consistent replacement
    words_to_ignore = {'struck', 'swinging', 'walked', 'grounded', 'into', 'double', 'play', 'second',
                       'shortstop', 'first', 'out', 'at', 'flied', 'to', 'center', 'right', 'left',
                       'pitcher', 'homered', 'feet', 'lined', 'popped', 'single', 'sacrifice', 'bunt',
                       'scored', 'hit', 'infield', 'reached', 'on', 'picked', 'off', 'caught', 'stealing',
                       'fielder\'s', 'choice', 'third', 'for', 'as', 'designated', 'hitter', 'new', 'player',
                       'base', 'looking', 'walk', 'runner', 'on', 'base', 'Game', 'Top', 'Bottom', 'Jr'}

    anonymization_mapping = {}
    pitcher_counter = 0
    player_counter = 0

    # Define a character class for letters (ASCII and common accented) compatible with re.UNICODE
    LETTER_CHARS_REGEX = r"[A-Za-z\u00C0-\u024F\u1E00-\u1EFF]"

    # Pass 1: Identify Pitchers explicitly
    all_pitcher_matches = re.findall(
        r'\b([A-Z]\.?\s?' + LETTER_CHARS_REGEX + r'+(?:[\s\-][A-Z]' + LETTER_CHARS_REGEX + r'+)*?)\s+PITCHING FOR (?:TEAM A|TEAM B)\.',
        anonymized_commentary, re.UNICODE
    )

    for p_name in sorted(list(set(all_pitcher_matches))):
        pitcher_counter += 1
        anonymization_mapping[p_name] = f"Pitcher {pitcher_counter}"

    # Pass 2: Identify all other names that appear as players
    player_name_pattern = re.compile(
        # Mc/Mac names
        r'\b(?:Mc|Mac)' + LETTER_CHARS_REGEX + r'+\'*(?:-' + LETTER_CHARS_REGEX + r'+\'*)?'
        # OR general names
        r'|[A-Z]' + LETTER_CHARS_REGEX + r'*(?:(?:[\s-][A-Z]' + LETTER_CHARS_REGEX + r'*)*)?'
        # OR Initial. Lastname
        r'|[A-Z]\.\s[A-Z]' + LETTER_CHARS_REGEX + r'*(?:(?:[\s-][A-Z]' + LETTER_CHARS_REGEX + r'*)*)?'
        r'(?:\sJr)?\b',
        re.UNICODE
    )

    for match in player_name_pattern.finditer(anonymized_commentary):
        original_name = match.group(0).strip()
        
        if len(original_name) <= 2 and not re.match(r'^[A-Z]\.$', original_name):
            continue
        if original_name.lower() in (w.lower() for w in words_to_ignore):
            continue
        if re.match(r'^\d+$', original_name):
            continue
        if re.match(r'^[A-Z]$', original_name) and not original_name.endswith('.'):
             continue

        if original_name in anonymization_mapping and anonymization_mapping[original_name].startswith("Pitcher"):
            continue

        if "TEAM A" in original_name or "TEAM B" in original_name:
            continue

        if original_name not in anonymization_mapping:
            player_counter += 1
            anonymization_mapping[original_name] = f"Player {player_counter}"

    sorted_original_names = sorted(anonymization_mapping.keys(), key=len, reverse=True)

    # 3. Perform Anonymization Replacements
    for original_name in sorted_original_names:
        escaped_name = re.escape(original_name)
        anonymized_commentary = re.sub(r'\b' + escaped_name + r'\b', anonymization_mapping[original_name], anonymized_commentary)
    
    # 4. Post-anonymization cleanup and formatting (line breaks)

    # Replace "Jr." with "Jr" for consistency
    anonymized_commentary = re.sub(r'Jr\.', 'Jr', anonymized_commentary)
    
    # Remove any extra spaces that might have resulted from replacements
    anonymized_commentary = re.sub(r'\s{2,}', ' ', anonymized_commentary)

    # Remove any stray dots from initials that might be left if not part of a name capture
    anonymized_commentary = re.sub(r'\b([A-Z])\.\s', r'\1 ', anonymized_commentary)

    # --- Apply desired line breaks ---

    # Add a double newline AFTER the initial "Game between..." line
    anonymized_commentary = re.sub(r'Game between (TEAM A|TEAM B) and (TEAM A|TEAM B)\.', r'Game between \1 and \2.\n\n', anonymized_commentary, 1)

    # Ensure a double newline BEFORE every inning marker (e.g., "--- Top 1st ---")
    # This also adds a single newline AFTER it.
    anonymized_commentary = re.sub(r'(?<!\n)\n?--- (Top|Bottom \d+th) ---', r'\n\n--- \1 ---\n', anonymized_commentary)

    # Add a single newline after each score update (e.g., "(TEAM A 0, TEAM B 0)")
    # This regex looks for a closing parenthesis followed by optional whitespace and then any character
    # that is NOT the start of "---" (to avoid adding a newline right before inning markers, as that's handled above)
    anonymized_commentary = re.sub(r'\)(\s*)(?!---)([A-Za-z0-9])', r')\n\2', anonymized_commentary)

    # Finally, ensure no more than two consecutive newlines anywhere, and strip overall leading/trailing whitespace.
    # This helps in cleaning up any accidental triples or more, while preserving doubles.
    anonymized_commentary = re.sub(r'\n{3,}', r'\n\n', anonymized_commentary).strip()
    
    return anonymized_commentary

# Provided commentary string
commentary = """Game between Diamondbacks and Reds.--- Top 1st ---B. SINGER PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Carroll struck out swinging. (Diamondbacks 0, Reds 0)Marte walked. (Diamondbacks 0, Reds 0)Perdomo grounded into double play, second to shortstop to first, Marte out at second. (Diamondbacks 0, Reds 0)--- Bottom 1st ---Z. GALLEN PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)Friedl flied out to center. (Diamondbacks 0, Reds 0)Lux flied out to center. (Diamondbacks 0, Reds 0)De La Cruz grounded out to second. (Diamondbacks 0, Reds 0)--- Top 2nd ---B. SINGER PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Naylor lined out to left. (Diamondbacks 0, Reds 0)Suárez homered to center (395 feet). (Diamondbacks 1, Reds 0)Gurriel Jr. struck out swinging. (Diamondbacks 1, Reds 0)Smith grounded out to pitcher. (Diamondbacks 1, Reds 0)--- Bottom 2nd ---Z. GALLEN PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)Encarnacion-Strand walked. (Diamondbacks 1, Reds 0)Benson flied out to center, Encarnacion-Strand to second. (Diamondbacks 1, Reds 0)Steer struck out swinging. (Diamondbacks 1, Reds 0)Trevino walked. (Diamondbacks 1, Reds 0)Fraley struck out swinging. (Diamondbacks 1, Reds 0)--- Top 3rd ---B. SINGER PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Thomas singled to right. (Diamondbacks 1, Reds 0)Herrera hit sacrifice bunt to pitcher, Thomas to second. (Diamondbacks 1, Reds 0)Carroll singled to center, Thomas scored. (Diamondbacks 2, Reds 0)Marte lined out to center. (Diamondbacks 2, Reds 0)Perdomo flied out to center. (Diamondbacks 2, Reds 0)--- Bottom 3rd ---Z. GALLEN PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)McLain singled to center. (Diamondbacks 2, Reds 0)Friedl grounded into double play, second to shortstop to first, McLain out at second. (Diamondbacks 2, Reds 0)Lux flied out to shortstop. (Diamondbacks 2, Reds 0)--- Top 4th ---B. SINGER PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Naylor grounded out to shortstop. (Diamondbacks 2, Reds 0)Suárez popped out to second. (Diamondbacks 2, Reds 0)Gurriel Jr. reached on infield single to pitcher. (Diamondbacks 2, Reds 0)Smith reached on infield single to second, Gurriel Jr. to second. (Diamondbacks 2, Reds 0)Thomas flied out to center. (Diamondbacks 2, Reds 0)--- Bottom 4th ---Z. GALLEN PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)De La Cruz lined out to right. (Diamondbacks 2, Reds 0)Encarnacion-Strand homered to left center (403 feet). (Diamondbacks 2, Reds 1)Benson grounded out to first. (Diamondbacks 2, Reds 1)Steer struck out looking. (Diamondbacks 2, Reds 1)--- Top 5th ---B. SINGER PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Herrera struck out looking. (Diamondbacks 2, Reds 1)Carroll walked. (Diamondbacks 2, Reds 1)Marte walked, Carroll to second. (Diamondbacks 2, Reds 1)Perdomo grounded into double play, second to shortstop to first, Marte out at second. (Diamondbacks 2, Reds 1)--- Bottom 5th ---Z. GALLEN PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)Trevino homered to left center (389 feet). (Diamondbacks 2, Reds 2)Fraley grounded out to shortstop. (Diamondbacks 2, Reds 2)McLain struck out looking. (Diamondbacks 2, Reds 2)Friedl flied out to right. (Diamondbacks 2, Reds 2)--- Top 6th ---S. BARLOW PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Naylor grounded out to first. (Diamondbacks 2, Reds 2)Suárez struck out swinging. (Diamondbacks 2, Reds 2)Gurriel Jr. lined out to center. (Diamondbacks 2, Reds 2)--- Bottom 6th ---Z. GALLEN PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)Lux struck out swinging. (Diamondbacks 2, Reds 2)De La Cruz walked. (Diamondbacks 2, Reds 2)Encarnacion-Strand singled to left, De La Cruz to second. (Diamondbacks 2, Reds 2)De La Cruz picked off and caught stealing third. (Diamondbacks 2, Reds 2)Benson grounded into fielder's choice to first, Encarnacion-Strand out at second. (Diamondbacks 2, Reds 2)--- Top 7th ---T. ROGERS PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Grichuk hit for Smith. (Diamondbacks 2, Reds 2)Grichuk struck out looking. (Diamondbacks 2, Reds 2)Thomas grounded out to shortstop. (Diamondbacks 2, Reds 2)Herrera struck out looking. (Diamondbacks 2, Reds 2)--- Bottom 7th ---Z. GALLEN PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)Grichuk as designated hitter. (Diamondbacks 2, Reds 2)Steer flied out to right. (Diamondbacks 2, Reds 2)Trevino flied out to left. (Diamondbacks 2, Reds 2)Fraley singled to right. (Diamondbacks 2, Reds 2)McLain homered to left (381 feet), Fraley scored. (Diamondbacks 2, Reds 4)J. BEEKS PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)Friedl flied out to right. (Diamondbacks 2, Reds 4)--- Top 8th ---T. ROGERS PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Carroll popped out to third. (Diamondbacks 2, Reds 4)Marte struck out swinging. (Diamondbacks 2, Reds 4)Perdomo flied out to center. (Diamondbacks 2, Reds 4)--- Bottom 8th ---K. BACKHUS PITCHING FOR ARI. (Diamondbacks ARI, Reds CIN)Lux popped out to shortstop. (Diamondbacks 2, Reds 4)De La Cruz struck out looking. (Diamondbacks 2, Reds 4)Encarnacion-Strand grounded out to third. (Diamondbacks 2, Reds 4)--- Top 9th ---T. SANTILLAN PITCHING FOR CIN. (Diamondbacks ARI, Reds CIN)Espinal at third base. (Diamondbacks 2, Reds 4)Naylor grounded out to first. (Diamondbacks 2, Reds 4)Suárez struck out swinging. (Diamondbacks 2, Reds 4)Gurriel Jr. grounded out to third. (Diamondbacks 2, Reds 4)"""

# Call the function to get the anonymized and formatted string
anonymized_commentary_output = anonymize_baseball_commentary(commentary)

# Print the resulting string
print("\n--- Original Commentary ---")
print(merged_df['english input'].iloc[0]) # Prints the original commentary for comparison
```
